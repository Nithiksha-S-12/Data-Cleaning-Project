```python
sns.boxplot(x=df['Fare'])
plt.show()
```

```python
Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df = df[(df['Fare'] >= lower) & (df['Fare'] <= upper)]
```

```python
df.shape
```

```python
df.drop(columns=['Cabin'])
```

```python
df.to_csv(r"C:\Users\Nithiksha S\OneDrive\Documents\Data-Cleaning-Visualization-Project\data\cleaned_data.csv.csv")
```

```python
sns.countplot(x='Survived', data=df)

plt.title("Survival Count")

plt.savefig(r"C:\Users\Nithiksha S\OneDrive\Documents\Data-Cleaning-Visualization-Project\images\survival_count.png")

plt.show()
```

Observation:
More passengers did not survive compared to those who survived.

```python
sns.barplot(x='Pclass', y='Survived', data=df)

plt.title("Passenger Class vs Survival")

plt.savefig(r"C:\Users\Nithiksha S\OneDrive\Documents\Data-Cleaning-Visualization-Project\images\class_survival.png")

plt.show()
```

Observation:
First-class passengers had a higher survival rate.
